```python
import torch 
import torch.nn as nn

vocab_size = 10000
d_model    = 512

embedding = nn.Embedding(vocab_size, d_model)
tokens    = torch.randint(0, vocab_size, (2, 10))

embeddings = embedding(tokens)
```

```python
print("Tokens shape:", tokens.shape)
print("Embeddings shape:", embeddings.shape)
```
